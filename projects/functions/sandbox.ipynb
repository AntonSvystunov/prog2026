```python
def add(first: int, second: int, times: int = 1) -> int:
    return (first + second) * times

print(add(2, 2, 3))
print(add(2, 2))
```

```python
def add(first: int, second: int, *, times: int = 1) -> int:
    return (first + second) * times

add(2, 2, times=3)
```

```python
def add(*, first: int, second: int, times: int = 1) -> int:
    return (first + second) * times

add(first=2, second=2, times=3)
```

```python
def add(numbers: list[int]) -> int:
    result = 0

    for number in numbers:
        result += number
    
    return result

print(add([1, 2]))
print(add([1, 2, 3]))
print(add([1, 2, 4]))

print(add([1]))
print(add([]))
```

```python
def add(first: int, second: int, *args: list[int]) -> int:
    result = first + second

    for number in args:
        result += number
    
    return result


print(add(1, 2, 3, 4, 5))
print(add(1, 2, 3, 4, 5, 7, 2))
print(add(1, 2))
```

```python
def add(first: int, second: int, *args: list[int], times: int = 1) -> int:
    result = first + second

    for number in args:
        result += number
    
    return result * times


print(add(1, 2, 3, 4, 5, times=2))
print(add(1, 2, 3, 4, 5, 7, 2))
print(add(1, 2))
```

```python
def say_hello(name: str, **kwargs: dict[str, bool]) -> None:
    print(f"Hello, {name}")

    if kwargs.get("ask_about_the_day"):
        print("How is your day")
    
    if kwargs.get("words"):
        print("Saying the words")



say_hello("Bob")
say_hello("John", ask_about_the_day=True, words=True)
```

```python
def divide(first: float, second: float) -> float:
    if second == 0:
        raise ValueError("Значення другого аргумента не може бути 0")

    return first / second

print(divide(9, 3))
print(divide(9, 2))
print(divide(9, 0))
```

```python
class NoAttemptsLeft(Exception):
    pass

def get_integer_input(*, prompt: str = "Введіть ціле число", attempts: int = 3) -> int:
    while attempts > 0:
        user_input = input(prompt)
        attempts -= 1

        try:
            return int(user_input)
        except:
            print("Помилка введення")
    
    raise NoAttemptsLeft("Використані всі спроби")


try:
    get_integer_input(prompt="Введіть номер групи:")
except NoAttemptsLeft:
    print("Використані всі спроби")
```

```python
import random

counts = {}
for _ in range(20_000):
    value = random.randint(1, 6)
    if value not in counts:
        counts[value] = 0
    counts[value] += 1

counts
```
