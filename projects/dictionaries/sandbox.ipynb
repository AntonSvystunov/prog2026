```python
with open("./text.txt", mode="r", encoding="utf-8") as file:
    text = file.read()
```

```python
import re

words = re.findall(r"[а-яА-ЯіІїЇґҐ']+", text)
words = [word.lower() for word in words]
```

```python
words_len = len(words)

words_count = {} 

for word in words:
    if word in words_count:
        words_count[word] = words_count[word] + 1 / words_len
    else:
        words_count[word] = 1 / words_len

words_count
```

```python
for words in words_count:
    print(words)
```

```python
for words in words_count.keys():
    print(words)
```

```python
for frequency in words_count.values():
    print(frequency)
```

```python
for pair in words_count.items():
    print(pair)
    word, frequency = pair
    print(f"{word}: {frequency:.2f}%")
```

```python
for word, frequency in words_count.items():
    print(f"{word}: {frequency:.2f}%")
```

```python
for word, frequency in list(sorted(words_count.items(), key=lambda pair: pair[1], reverse=True))[0:5]:
    print(f"{word}: {frequency:.2f}%")
```
